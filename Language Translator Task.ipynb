```python
from tkinter import * 
from tkinter import ttk
from deep_translator import GoogleTranslator
```

```python
#crete a main window
root=Tk()
root.geometry('600x350')
root.resizable(0,0)
#root.iconbitmap('logo translator.ico')
root['bg']='White'
#create title label
root.title('Language Translator')
Label(root,text="Language translator",font=("Arial",16,"bold"),bg='white').pack() 
```

```python
#Create Text Box
Label(root,text="Enter Text",font="Arial,16,bold",bg='white smoke').place(x=50,y=60)
```

```python
Input_text=Text(root,width=30,height=5)
Input_text.place(x=50,y=90)
```

```python
Label(root,text="Output",font='arial 16 bold',bg='white smoke').place(x=330,y=60)
output_text=Text(root,font='arial 10',height=5, wrap=WORD,padx=5,pady=5,width=30)
output_text.place(x=330,y=90)
```

```python
LANGUAGES=GoogleTranslator().get_supported_languages()
language=list(LANGUAGES)
dest_lang=ttk.Combobox(root,values=language,width=25)
dest_lang.place(x=50,y=190)
dest_lang.set('Choose Language')

def translate_text():
    text = Input_text.get("1.0",END)
    lang=dest_lang.get()
    if text.strip() == ""or lang =="choose language":
        output_text.delete("1.0",END)
        output_text.insert(END,"Enter text & select language")
        return
    translated=GoogleTranslator(source='auto',target=lang).translate(text)
    output_text.delete("1.0",END)
    output_text.insert(END,translated)
def reset_text():    
    Input_text.delete("1.0",END)
    output_text.delete("1.0",END)
    dest_lang.set("Choose language")

Button(root,text="Translate",command=translate_text).place(x=180,y=270)    
Button(root,text="Reset",command=reset_text).place(x=320,y=270)
root.mainloop()
```
